## Baseline

# M1

```python
import glob

parquet_files = glob.glob(dataFolder + "*.parquet")

raw_trips = spark.read.parquet(*parquet_files)

#checking the schema
raw_trips.printSchema()
```

```python
import pyspark.sql.functions as sf

# enforcing schema
trips = (raw_trips
    .withColumn("tpep_pickup_datetime", sf.col("tpep_pickup_datetime").cast("timestamp")) # (col creat/overwr , opp)
    .withColumn("tpep_dropoff_datetime", sf.col("tpep_dropoff_datetime").cast("timestamp"))
    .withColumn("PULocationID", sf.col("PULocationID").cast("integer"))
    .withColumn("DOLocationID", sf.col("DOLocationID").cast("integer"))
    .withColumn("passenger_count", sf.col("passenger_count").cast("integer"))
    .withColumn("trip_distance", sf.col("trip_distance").cast("double"))
    .withColumn("fare_amount", sf.col("fare_amount").cast("double"))
    .withColumn("year", sf.year(sf.col("tpep_pickup_datetime")))
)

# handling the broken data - vars are just holding condition
invalid_fare      = (sf.col("fare_amount").isNull()) | (sf.col("fare_amount") <= 0)
invalid_distance  = (sf.col("trip_distance").isNull()) | (sf.col("trip_distance") <= 0)
invalid_passengers= (sf.col("passenger_count").isNull()) | (sf.col("passenger_count") == 0) | (sf.col("passenger_count") > 6)
invalid_timestamps= (sf.col("tpep_pickup_datetime").isNull()) | (sf.col("tpep_dropoff_datetime").isNull()) | (sf.col("tpep_dropoff_datetime") <= sf.col("tpep_pickup_datetime"))
invalid_locations = (sf.col("PULocationID").isNull()) | (sf.col("DOLocationID").isNull()) | \
                    (~sf.col("PULocationID").between(1, 263)) | (~sf.col("DOLocationID").between(1, 263))

# combining conditions to one cond
invalid_any = invalid_fare | invalid_distance | invalid_passengers | invalid_timestamps | invalid_locations

# 4. Compute and report per-year cleaning statistics (count of records removed per criterion)
# We cast boolean flags to integers (True -> 1, False -> 0) so sum() acts as a counter
cleaning_stats = (trips
    .groupBy("year")
    .agg(
        sf.count("*").alias("total_records_processed"),
        sf.sum(invalid_fare.cast("integer")).alias("removed_invalid_fare"), #for each year applies the statement
        #and each row assigned with boolean, but then we cast integer, and we sum,
        # so it creates new dataframe with year and number
        sf.sum(invalid_distance.cast("integer")).alias("removed_invalid_distance"),
        sf.sum(invalid_passengers.cast("integer")).alias("removed_invalid_passengers"),
        sf.sum(invalid_timestamps.cast("integer")).alias("removed_invalid_timestamps"),
        sf.sum(invalid_locations.cast("integer")).alias("removed_invalid_locations"),
        sf.sum(invalid_any.cast("integer")).alias("total_records_removed")
    )
    .orderBy("year")
)

# we filter from main trips dataset
m1_clean_trips = trips.filter(~invalid_any)

print("--- M1 Data Cleaning Statistics Report ---")
cleaning_stats.show(truncate=False)
```

```python
m1_clean_trips.printSchema()
```

# M2

```python
import os

zone_lookup_path = dataFolder + "taxi_zone_lookup.csv"

#loading just 2 cols
zone_lookup = (spark.read
    .option("header", "true")
    .option("inferSchema", "true")
    .csv(zone_lookup_path)
    .select(
        sf.col("LocationID").cast("integer").alias("zone_id"),
        sf.col("Borough").alias("pickup_borough")
    )
)

m2_prepared_trips = (m1_clean_trips
    .withColumn("day_of_week", sf.dayofweek(sf.col("tpep_pickup_datetime")))  
    .withColumn("hour_of_day", sf.hour(sf.col("tpep_pickup_datetime")))
    .withColumn("duration_minutes", 
        (sf.unix_timestamp(sf.col("tpep_dropoff_datetime")) - sf.unix_timestamp(sf.col("tpep_pickup_datetime"))) / 60.0
    )
)

#for optimization performing the broadcast join - avoiding shuffle and makes sense because csv small
#
m2_joined = m2_prepared_trips.join(
    zone_lookup,
    m2_prepared_trips.PULocationID == zone_lookup.zone_id,
    how="inner"
)


m2_demand_analysis = (m2_joined
    .groupBy("pickup_borough", "day_of_week", "hour_of_day")
    .agg(
        sf.count("*").alias("total_trip_count"),
        sf.mean("fare_amount").alias("mean_fare_amount"),
        sf.mean("duration_minutes").alias("mean_trip_duration_min"),
        sf.mean("trip_distance").alias("mean_trip_distance")
    )
    .orderBy("pickup_borough", "day_of_week", "hour_of_day")
)

output_path_m2 = "/tmp/m2_temporal_demand_output"

spark.sparkContext.setJobGroup("M2_ANALYSIS", "M2: Temporal Demand Parquet Write") #for spark ui

(m2_demand_analysis
    .write
    .mode("overwrite")
    .partitionBy("pickup_borough")
    .parquet(output_path_m2)
)

spark.sparkContext.setJobGroup("M2_ANALYSIS", "show 10 lines") 

print("--- M2 Temporal Demand Analysis 10 first ---")
m2_demand_analysis.show(10, truncate=False)

spark.sparkContext.setJobGroup(" ", " ") # cleaning labeling for sparkui
```

# M3

```python
from pyspark.sql.window import Window

spark.sparkContext.setJobGroup("M3_STATS", "M3: Zone-Level Window Statistics Calculation")

m3_prepared = (m1_clean_trips
    .withColumn("year_month", sf.date_format(sf.col("tpep_pickup_datetime"), "yyyy-MM"))
    .join(
        zone_lookup,
        m1_clean_trips.PULocationID == zone_lookup.zone_id,
        how="inner"
    )
    .filter(sf.col("pickup_borough").isNotNull())
)


m3_monthly_zone_stats = (m3_prepared
    .groupBy("pickup_borough", "PULocationID", "year_month") #we need pickup borough for laters window
    .agg(
        sf.count("*").alias("monthly_trip_count"),
        sf.mean("fare_amount").alias("mean_fare_amount"),
        sf.stddev("fare_amount").alias("std_fare_amount"),
        sf.mean("trip_distance").alias("mean_trip_distance"),
        sf.stddev("trip_distance").alias("std_trip_distance")
    )
)

# it assigns to the buckets, we dont collapse rows here like in a group by
w_borough_monthly_rank = Window.partitionBy("pickup_borough", "year_month").orderBy(sf.col("monthly_trip_count").desc())

w_borough_monthly_total = Window.partitionBy("pickup_borough", "year_month")

w_zone_rolling_3m = (Window
    .partitionBy("PULocationID")
    .orderBy("year_month")
    .rowsBetween(-2, 0)
)


m3_final = (m3_monthly_zone_stats
    
    .withColumn("zone_volume_rank", sf.rank().over(w_borough_monthly_rank)) #here the index is created
    
    # for each neigbourhood and month, duplicated the same value of total FOR EACH NEIGH MONTLY TRIPS COUNT
    .withColumn("borough_monthly_trips", sf.sum("monthly_trip_count").over(w_borough_monthly_total)) 

    # each id's total count over the neighberhoods total
    .withColumn("zone_monthly_trip_share", sf.col("monthly_trip_count") / sf.col("borough_monthly_trips"))
    
    .withColumn("rolling_3m_trip_share", sf.sum("zone_monthly_trip_share").over(w_zone_rolling_3m))
    
    # avg and std for rolling 3 months
    .withColumn("rolling_volume_mean", sf.avg("monthly_trip_count").over(w_zone_rolling_3m))
    .withColumn("rolling_volume_std", sf.stddev("monthly_trip_count").over(w_zone_rolling_3m))
    
    # Flag zones where |current volume - rolling mean| > 1.5 * rolling std
    .withColumn(
        "is_volume_anomaly",
        sf.when(
            sf.col("rolling_volume_std").isNotNull() & (sf.col("rolling_volume_std") > 0),
            sf.abs(sf.col("monthly_trip_count") - sf.col("rolling_volume_mean")) > (1.5 * sf.col("rolling_volume_std"))
        ).otherwise(False)
    )
)


output_path_m3 = "/tmp/m3_zone_window_stats_output"

(m3_final
    .write
    .mode("overwrite")
    .partitionBy("pickup_borough")
    .parquet(output_path_m3)
)

print("--- M3 Zone-Level Window Statistics (First 10 Rows) ---")
m3_final.show(10, truncate=False)

# reseting the name 
spark.sparkContext.setJobGroup("", "")
```

# O1

```python
spark.sparkContext.setJobGroup("O1_ANOMALY_BASELINE", "O1: Fare Anomaly Detection Baseline")

with profiler.profile("O1-fare-anomaly-baseline"): #for logs
    
    # extracting the fare , hour and weekend (later weekday vs weekend)
    o1_base = (m1_clean_trips
        .filter(sf.col("trip_distance") > 0)
        .withColumn("fare_per_dist", sf.col("fare_amount") / sf.col("trip_distance"))
        .withColumn("hour_of_day", sf.hour(sf.col("tpep_pickup_datetime")))
        .withColumn("is_weekend", sf.dayofweek(sf.col("tpep_pickup_datetime")).isin([1, 7]))
    )

    # mean and stddev
    o1_stats = (o1_base
        .groupBy("PULocationID", "hour_of_day")
        .agg(
            sf.mean("fare_per_dist").alias("mean_fare_per_dist"),
            sf.stddev("fare_per_dist").alias("std_fare_per_dist")
        )
    )

    # 3. Join stats back to label trips exceeding mean + 3*stddev
    o1_flagged = (o1_base
        .join(o1_stats, on=["PULocationID", "hour_of_day"], how="inner")
        .join( #take the neighberhood
            zone_lookup.select(
                sf.col("zone_id").alias("PULocationID"),
                "pickup_borough"
            ),
            on="PULocationID",
            how="inner"
        )
        .withColumn( #add the flag
            "is_fare_anomaly",
            sf.when(
                sf.col("std_fare_per_dist").isNotNull() & (sf.col("std_fare_per_dist") > 0),
                sf.col("fare_per_dist") > (sf.col("mean_fare_per_dist") + (3 * sf.col("std_fare_per_dist"))) 
            ).otherwise(False)
        )
    )

    # 4. Aggregate weekday vs. weekend anomaly rates by payment_type and borough
    o1_summary = (o1_flagged
        .groupBy("pickup_borough", "payment_type", "is_weekend")
        .agg(
            sf.count("*").alias("total_trips"),
            sf.sum(sf.when(sf.col("is_fare_anomaly"), 1).otherwise(0)).alias("anomaly_trips"), # True if anomaly
            (sf.sum(sf.when(sf.col("is_fare_anomaly"), 1).otherwise(0)) / sf.count("*")).alias("anomaly_rate") 
        ) #perc. of total anomalies for that group
        .orderBy("pickup_borough", "payment_type", "is_weekend")
    )

    print("--- O1 Fare Anomaly Summary (First 10 Rows) ---")
    o1_summary.show(10, truncate=False)

spark.sparkContext.setJobGroup("", "")

#for optimizing
#1 broadcastjoin
#2 getting rid of 1st join completely
```

# O2

```python
spark.sparkContext.setJobGroup("O2_FLOWS_BASELINE", "O2: Origin-Destination Flow Analysis Baseline")

with profiler.profile("O2-origin-destination-baseline"):

    #base with trip counts
    o2_pairs = (m1_clean_trips
        .withColumn("year_month", sf.date_format(sf.col("tpep_pickup_datetime"), "yyyy-MM"))
        .groupBy("PULocationID", "DOLocationID", "year_month")
        .agg(sf.count("*").alias("pair_trip_count"))
    )

    zone_lookup_full = (spark.read
        .option("header", "true")
        .option("inferSchema", "true")
        .csv(dataFolder + "taxi_zone_lookup.csv")
        .select(
            sf.col("LocationID").cast("integer").alias("zone_id"),
            sf.col("Borough").alias("borough"),
            sf.col("Zone").alias("zone_name")
        )
    )

    zone_pickup = zone_lookup_full.select(
        sf.col("zone_id").alias("PULocationID"),
        sf.col("borough").alias("pickup_borough"),
        sf.col("zone_name").alias("pickup_zone")
    )
    zone_dropoff = zone_lookup_full.select(
        sf.col("zone_id").alias("DOLocationID"),
        sf.col("borough").alias("dropoff_borough"),
        sf.col("zone_name").alias("dropoff_zone")
    )
 
    o2_named = (o2_pairs
        .join(zone_pickup, on="PULocationID", how="inner")
        .join(zone_dropoff, on="DOLocationID", how="inner")
    )

    w_pickup_month = Window.partitionBy("PULocationID", "year_month")
    
    o2_shares = (o2_named
        .withColumn("total_outbound_trips", sf.sum("pair_trip_count").over(w_pickup_month)) #each pul month total trips
        .withColumn("route_outbound_share", sf.col("pair_trip_count") / sf.col("total_outbound_trips"))
    )

    w_borough_rank = Window.partitionBy("pickup_borough", "year_month").orderBy(sf.col("pair_trip_count").desc())
    
    o2_top20 = (o2_shares
        .withColumn("borough_route_rank", sf.rank().over(w_borough_rank))
        .filter(sf.col("borough_route_rank") <= 20)
    )

    o2_quarters = (o2_top20
        .withColumn("quarter", sf.quarter(sf.to_date(sf.col("year_month"), "yyyy-MM"))) 
        .filter(sf.col("quarter").isin([1, 3])) 
        .groupBy("PULocationID", "DOLocationID", "pickup_borough", "pickup_zone", "dropoff_zone", "quarter")
        .agg(sf.avg("route_outbound_share").alias("avg_quarter_share"))
    )

    o2_q1 = o2_quarters.filter(sf.col("quarter") == 1).alias("q1")
    o2_q3 = o2_quarters.filter(sf.col("quarter") == 3).alias("q3")

    o2_reversals = (o2_q1
        .join(
            o2_q3,
            (sf.col("q1.PULocationID") == sf.col("q3.PULocationID")) &
            (sf.col("q1.DOLocationID") == sf.col("q3.DOLocationID")),
            how="inner"
        )
        .withColumn("share_delta", sf.col("q3.avg_quarter_share") - sf.col("q1.avg_quarter_share"))
        .select(
            sf.col("q1.pickup_borough"),
            sf.col("q1.pickup_zone"),
            sf.col("q1.dropoff_zone"),
            sf.col("q1.avg_quarter_share").alias("q1_share"),
            sf.col("q3.avg_quarter_share").alias("q3_share"),
            "share_delta"
        )
        .orderBy(sf.abs(sf.col("share_delta")).desc()) 
    )

    print("--- O2 Top 10 Route Share Reversals (Q1 vs Q3) ---")
    o2_reversals.show(10, truncate=False)

spark.sparkContext.setJobGroup("", "")
```

# O1 Optimization

```python
# Enable Adaptive Query Execution (AQE)
# its for the dataskew - splitting heavy partitions and merging small partitions together

spark.conf.set("spark.sql.adaptive.enabled", "true")
spark.conf.set("spark.sql.adaptive.coalescePartitions.enabled", "true")
spark.conf.set("spark.sql.adaptive.skewJoin.enabled", "true")
spark.conf.set("spark.sql.adaptive.localShuffleReader.enabled", "true")
spark.conf.set("spark.sql.shuffle.partitions", "32")
```

```python
#optimized with using the broadcast join and AQE
spark.sparkContext.setJobGroup("O1_ANOMALY_OPTIMIZED", "O1: Fare Anomaly Detection Optimized")

with profiler.profile("O1-fare-anomaly-optimized"):
    #creating the base
    o1_base_opt = (m1_clean_trips
        .filter(sf.col("trip_distance") > 0)
        .withColumn("fare_per_dist", sf.col("fare_amount") / sf.col("trip_distance"))
        .withColumn("hour_of_day", sf.hour(sf.col("tpep_pickup_datetime")))
        .withColumn("is_weekend", sf.dayofweek(sf.col("tpep_pickup_datetime")).isin([1, 7]))
    )

    o1_stats_opt = (o1_base_opt
        .groupBy("PULocationID", "hour_of_day")
        .agg(
            sf.mean("fare_per_dist").alias("mean_fare_per_dist"),
            sf.stddev("fare_per_dist").alias("std_fare_per_dist")
        )
    )
    #broadcasts joins 
    o1_flagged_opt = (o1_base_opt
        .join(
            sf.broadcast(o1_stats_opt), 
            on=["PULocationID", "hour_of_day"], 
            how="inner"
        )
        .join(
            sf.broadcast(
                zone_lookup.select(
                    sf.col("zone_id").alias("PULocationID"),
                    "pickup_borough"
                )
            ),
            on="PULocationID",
            how="inner"
        )
        .withColumn(
            "is_fare_anomaly",
            sf.when(
                sf.col("std_fare_per_dist").isNotNull() & (sf.col("std_fare_per_dist") > 0),
                sf.col("fare_per_dist") > (sf.col("mean_fare_per_dist") + (3 * sf.col("std_fare_per_dist")))
            ).otherwise(False)
        )
    )
 
    o1_summary_opt = (o1_flagged_opt 
        .groupBy("pickup_borough", "payment_type", "is_weekend")
        .agg(
            sf.count("*").alias("total_trips"),
            sf.sum(sf.when(sf.col("is_fare_anomaly"), 1).otherwise(0)).alias("anomaly_trips"), #true if anomaly
            (sf.sum(sf.when(sf.col("is_fare_anomaly"), 1).otherwise(0)) / sf.count("*")).alias("anomaly_rate") 
        )
        .orderBy("pickup_borough", "payment_type", "is_weekend")
    )

    print("--- O1 Fare Anomaly Summary (Optimized) ---")
    o1_summary_opt.show(10, truncate=False)

print(profiler.compare("O1-fare-anomaly-baseline", "O1-fare-anomaly-optimized"))
spark.sparkContext.setJobGroup("", "")
```

```python
#FAIL WITH WINDOW
#mention in raport that thought that avoiding join with window will benefit but didnot

spark.sparkContext.setJobGroup("O1_ANOMALY_OPTIMIZED", "O1: Fare Anomaly Detection Optimized")

#plan
#broadcast join for zone lookup
#calc on window to avoid the shuffle

with profiler.profile("O1-fare-anomaly-optimized"):
    
    w_bucket = Window.partitionBy("PULocationID", "hour_of_day")

    # 1. Feature extraction + Inline Window Statistics + Anomaly Flagging in a single pass
    o1_flagged_opt = (m1_clean_trips
        .filter(sf.col("trip_distance") > 0)
        .withColumn("fare_per_dist", sf.col("fare_amount") / sf.col("trip_distance"))
        .withColumn("hour_of_day", sf.hour(sf.col("tpep_pickup_datetime")))
        .withColumn("is_weekend", sf.dayofweek(sf.col("tpep_pickup_datetime")).isin([1, 7])) #base
        
        #calculating here to avoid shuffle join
        .withColumn("mean_fare_per_dist", sf.mean("fare_per_dist").over(w_bucket)) #for each pulid and hour
        .withColumn("std_fare_per_dist", sf.stddev("fare_per_dist").over(w_bucket)) 
        
        .withColumn(
            "is_fare_anomaly", #flagging
            sf.when(
                sf.col("std_fare_per_dist").isNotNull() & (sf.col("std_fare_per_dist") > 0), #1st arg
                sf.col("fare_per_dist") > (sf.col("mean_fare_per_dist") + (3 * sf.col("std_fare_per_dist"))) #condit
            ).otherwise(False)
        )

        #broadcasting zonelookup table
        .join(
            sf.broadcast(
                zone_lookup.select(
                    sf.col("zone_id").alias("PULocationID"),
                    "pickup_borough"
                )
            ),
            on="PULocationID",
            how="inner"
        )
    )

    o1_summary_opt = (o1_flagged_opt
        .groupBy("pickup_borough", "payment_type", "is_weekend")
        .agg(
            sf.count("*").alias("total_trips"),
            sf.sum(sf.when(sf.col("is_fare_anomaly"), 1).otherwise(0)).alias("anomaly_trips"), #true if anomaly
            (sf.sum(sf.when(sf.col("is_fare_anomaly"), 1).otherwise(0)) / sf.count("*")).alias("anomaly_rate") 
        )
        .orderBy("pickup_borough", "payment_type", "is_weekend")
    )

    print("--- O1 Fare Anomaly Summary (Optimized) ---")
    o1_summary_opt.show(10, truncate=False)

print(profiler.compare("O1-fare-anomaly-baseline", "O1-fare-anomaly-optimized"))
spark.sparkContext.setJobGroup("", "")
```

# O2 Optimization

```python
spark.sparkContext.setJobGroup("O2_FLOWS_OPTIMIZED", "O2: Origin-Destination Flow Analysis Optimized")

with profiler.profile("O2-origin-destination-optimized"):
    
   # pair with trips count
    o2_pairs_opt = (m1_clean_trips
        .withColumn("year_month", sf.date_format(sf.col("tpep_pickup_datetime"), "yyyy-MM"))
        .groupBy("PULocationID", "DOLocationID", "year_month")
        .agg(sf.count("*").alias("pair_trip_count"))
    )

    #broadcast join only the borough needed for ranking
    zone_borough_bc = sf.broadcast(
        zone_lookup_full.select(
            sf.col("zone_id").alias("PULocationID"), 
            sf.col("borough").alias("pickup_borough")
        )
    )
    
    o2_with_borough = o2_pairs_opt.join(zone_borough_bc, on="PULocationID", how="inner")

    w_pickup_month = Window.partitionBy("PULocationID", "year_month")
    w_borough_rank = Window.partitionBy("pickup_borough", "year_month").orderBy(sf.col("pair_trip_count").desc())

    o2_top20_opt = (o2_with_borough #
        .withColumn("total_outbound_trips", sf.sum("pair_trip_count").over(w_pickup_month)) #each loc and month total trips
        .withColumn("route_outbound_share", sf.col("pair_trip_count") / sf.col("total_outbound_trips")) 
        .withColumn("borough_route_rank", sf.rank().over(w_borough_rank))
        .filter(sf.col("borough_route_rank") <= 20) # early filtering
    )

    o2_reversals_raw = (o2_top20_opt
        .withColumn("quarter", sf.quarter(sf.to_date(sf.col("year_month"), "yyyy-MM")))
        .filter(sf.col("quarter").isin([1, 3]))
        .groupBy("PULocationID", "DOLocationID", "pickup_borough")
        .agg(
            sf.avg(sf.when(sf.col("quarter") == 1, sf.col("route_outbound_share"))).alias("q1_share"),
            sf.avg(sf.when(sf.col("quarter") == 3, sf.col("route_outbound_share"))).alias("q3_share")
        )
        .filter(sf.col("q1_share").isNotNull() & sf.col("q3_share").isNotNull())
        .withColumn("share_delta", sf.col("q3_share") - sf.col("q1_share"))
    )

# braodcast join
    zone_p_bc = sf.broadcast(
        zone_lookup_full.select(
            sf.col("zone_id").alias("PULocationID"),
            sf.col("zone_name").alias("pickup_zone")
        )
    )
    zone_d_bc = sf.broadcast(
        zone_lookup_full.select(
            sf.col("zone_id").alias("DOLocationID"),
            sf.col("zone_name").alias("dropoff_zone")
        )
    )

    o2_reversals_opt = (o2_reversals_raw
        .join(zone_p_bc, on="PULocationID", how="inner")
        .join(zone_d_bc, on="DOLocationID", how="inner")
        .select("pickup_borough", "pickup_zone", "dropoff_zone", "q1_share", "q3_share", "share_delta")
        .orderBy(sf.abs(sf.col("share_delta")).desc())
    )

    print("--- O2 Top 10 Route Share Reversals (Optimized) ---")
    o2_reversals_opt.show(10, truncate=False)

print(profiler.compare("O2-origin-destination-baseline", "O2-origin-destination-optimized"))
spark.sparkContext.setJobGroup("", "")
```
